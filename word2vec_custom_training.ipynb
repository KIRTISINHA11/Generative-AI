```python
#end goal of text preprocessing is to convert text into tokens
folder_path = r"C:\Users\Admin\Agentic AI Projects\AI\Machine Learning\data"
story =[]
for file in os.listdir(folder_path):    
    if file.endswith('.txt'):
        with open(os.path.join(folder_path,file),'r',encoding='utf-8') as f:
            corpus = f.read()
#tokenize the data
raw_sentences = sent_tokenize(corpus)
for sentence in raw_sentences:
    story.append(simple_preprocess(sentence))

```

```python
story
```

```python
story[0]
```

```python
custom_model =gensim.models.Word2Vec(
    window=10,
    min_count=1,   #reduce noise, very rare data not sent for training
    vector_size=150  #dimension of vector , google model has 300 dimensions, can keep anything # no. of neurons in hidden layer
    
)
#by default it uses skip gram 
#sg=0 means cbow
#sg=1 means skip gram
#less data cbow
#more data skip gram
#hyperparameter tuning can be done by changing the parameters of the model
```

```python
custom_model.build_vocab(story)
```

```python
custom_model.corpus_count,# total no. of sentences in corpus/data
```

```python
custom_model.train(story, total_examples=custom_model.corpus_count, epochs=10)
#epoch hyperparameter , can be anything, experimental
```
